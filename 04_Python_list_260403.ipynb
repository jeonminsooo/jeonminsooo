```python
# 1 ~ N의 수로 이루어진 리스트 생성
N = int(input("숫자 : "))
num_list = []
for i in range(1,N+1):
    num_list.append(i)
print(num_list)
```

```python
# 1 ~ N의 수로 이루어진 리스트 생성
N = int(input("숫자 : "))
num_list = [i for i in range(1,N+1)]
print(num_list)

# expression
num_list = [i * 2 for i in range(1,N+1)]
print(num_list)

# 1 ~ N의 수로 이루어진 리스트 생성
N = int(input("숫자 : "))
num_list = [i for i in range(1,N+1) if i % 2 == 0]
print(num_list)
```
